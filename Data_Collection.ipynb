```python
# Web Scraping
import requests
URL= "https://www.scrapethissite.com/pages/simple/"
res = requests.get(URL)
print(res.status_code)
```

```python
if res.status_code == 200:
    print(res.headers)
    # print(res.content)
with open ("scraped_data/data.html","w") as f:
    f.write(res.text)
```

```python
# Beautiful soup
!pip install beautifulsoup4
from bs4 import BeautifulSoup
with open("scraped_data/data.html","r") as f:
    html_content = f.read()
soup = BeautifulSoup(html_content,"lxml")
```

```python
soup.find("h1")
allh3 = soup.find_all("h3")
all_countries = []
type(allh3)
for h3 in allh3:
    name = (h3.get_text(strip = True))
    # print(h3.find_parent("div")["class"])
    population = (h3.find_next("div").select("span.country-population")[0].get_text(strip = True))
    all_countries.append([name,population])
```

```python
import pandas as pd
df = pd.DataFrame(all_countries,columns = ["Name","Polpulation"])
```

```python
df.to_csv("Cleaned_data/data.csv",index = False)
```

```python
#Assignment problem - Activity
import requests
from bs4 import BeautifulSoup
i = 1
while True:
    URL = f"https://quotes.toscrape.com/page/{i}/"
    res = requests.get(URL)
    soup = BeautifulSoup(res.text,"lxml")
    quotes = soup.select("div.quote")
    if not quotes:
        print("No valid Pages anymore...")
        break
    with open(f"scraped_data/data{i}.html","w") as f:
        f.write(res.text)
        print(f"downloaded data from page{i}")
    i=i+1
```

```python
#Extrct Info
life_quotes = []
for i in range(1,11):
    with open(f"scraped_data/data{i}.html","r") as f:
        html_content = f.read()
    soup = BeautifulSoup(html_content,"lxml")
    all_quotes = soup.select("div.quote")
    # print(all_quotes)
    for q in all_quotes:
        all_tags = []
        for tag in q.select(".tags .tag"):
            all_tags.append(tag.get_text())
        # print(all_tags)
        if "life" in all_tags:
            text = q.select_one("span.text").get_text()
            # print(text)
            author = q.select_one("small.author").get_text()
            # print(author)
            life_quotes.append([text,author])
print(life_quotes)
```
