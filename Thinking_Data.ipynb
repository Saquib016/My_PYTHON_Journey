```python
import json
```

```python
# Load the data

def load_data(filename):
    with open(filename,"r") as f:
        data = json.load(f)
    return data
```

```python
data = load_data("store_data.json")
print(data)
print(type(data))
```

```python
# clean and struncured the data
def clean_data(data):
    text_to_num = {"one":1,"two": 2,"three":3,"four":4,"five":5}
    cleaned_data = []
    unique_user = set()
    for user in data:
        raw_rating = user["rating"].strip().lower()
        if(raw_rating in text_to_num):
            raw_rating  = text_to_num[raw_rating]
        user["rating"] = (raw_rating)
        #handle missing
        raw_age = user.get("age")
        if(raw_age==None):
            user["age"] = None
        #Deduplication
        if(user["name"].strip() in unique_user):
            continue
        unique_user.add(user["name"])
        cleaned_data.append(user)
    return cleaned_data
```

```python
#get meaningful Insights
def get_insight(data):
    tot_rating = 0;
    for user in data:
        tot_rating += float(user["rating"])
    print(f"average rating = {tot_rating/len(data)}")
    #percentage of users with poor rating
    poor = 0
    for user in data:
        if(float(user["rating"])<3):
            poor+=1
            print(user["name"])
    print(f"Percentager of user give bad Rating = {poor/len(data)*100}")
```

```python
clean_data(data)
```

```python
get_insight(data)
```

```python
#Recommendation system
def get_recommendation(data):
    recommendations = []
    for user in data:
        curr_recom = {}
        curr_recom["name"] = user["name"]
        if(float(user["rating"])>=4):
            curr_recom["brand"]  ="Apple"
        else:
            curr_recom["brand"] = "Samsung"
        recommendations.append(curr_recom)
    return recommendations
```

```python
get_recommendation(data)
```
